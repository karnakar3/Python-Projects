```python
##importing random to get a random number on the dice

import random

## function definition
def roll_dice():
    return random.randint(1,6)

## driver code

## initializing the dice faces (1-6) using dictionary
dice_faces={
    1:(" -------- \n"
       "|        |\n"
       "|   ()   |\n"
       "|        |\n"
       " -------- \n"),
    2:(" -------- \n"
       "| ()     |\n"
       "|        |\n"
       "|     () |\n"
       " -------- \n"),
    3:(" -------- \n"
       "| ()     |\n"
       "|   ()   |\n"
       "|     () |\n"
       " -------- \n"),
    4:(" -------- \n"
       "| ()  () |\n"
       "|        |\n"
       "| ()  () |\n"
       " -------- \n"),
    5:(" -------- \n"
       "| ()  () |\n"
       "|   ()   |\n"
       "| ()  () |\n"
       " -------- \n"),
    6:(" -------- \n"
       "| ()  () |\n"
       "| ()  () |\n"
       "| ()  () |\n"
       " -------- \n")
}

print("welcome to dice rolling simulator\n")
## using loop to roll the dice multiple times
play_again='y'
while play_again =='y':
    input("press enter to roll the dice...")
    ## function calling
    result=roll_dice()
    print(dice_faces[result])
    
    ## input for playing again
    play_again=input("do you want to play again ? (y/n):")
                
print()        
print("thanks for playing")
```
